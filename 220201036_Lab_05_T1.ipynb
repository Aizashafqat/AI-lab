```python
import math

def minimax(depth, node_index, is_max, values, alpha, beta):
    if depth == 2:
        return values[node_index]
    
    if is_max:
        best = -math.inf
        for i in range(2):
            val = minimax(depth + 1, node_index * 2 + i, False, values, alpha, beta)
            best = max(best, val)
            alpha = max(alpha, best)
            if beta <= alpha:
                break
        return best
    
    else:
        best = math.inf
        for i in range(2):
            val = minimax(depth + 1, node_index * 2 + i, True, values, alpha, beta)
            best = min(best, val)
            beta = min(beta, best)
            if beta <= alpha:
                break
        return best

values = [10, 15, 5, 20]
optimal_value = minimax(0, 0, True, values, -math.inf, math.inf)

print(f"Optimal Value for MAX: {optimal_value}")
```
