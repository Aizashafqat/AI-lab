```python
import math

def minimax(depth, node_index, is_max, values, alpha, beta):
    if depth == 2:
        return values[node_index]
    
    if is_max:
        best = -math.inf
        for i in range(2):
            val = minimax(depth + 1, node_index * 2 + i, False, values, alpha, beta)
            best = max(best, val)
            alpha = max(alpha, best)
            if beta <= alpha:
                print(f"Pruning at MAX node: α={alpha}, β={beta}")
                break
        return best
    
    else:
        best = math.inf
        for i in range(3):  # Right MIN node has 3 children
            val = minimax(depth + 1, node_index * 3 + i, True, values, alpha, beta)
            best = min(best, val)
            beta = min(beta, best)
            if beta <= alpha:
                print(f"Pruning at MIN node with [{values[node_index * 3 + i]}]")
                break
        return best

values = [8, 12, 4, 6, 9, 3]
optimal_value = minimax(0, 0, True, values, -math.inf, math.inf)

print(f"\nOptimal Value for MAX: {optimal_value}")
```
