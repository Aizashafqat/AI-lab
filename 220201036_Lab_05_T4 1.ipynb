```python
import math

def minimax(depth, node_index, is_max, values, alpha, beta, structure):
    if depth == 2:
        return values[node_index]
    
    if is_max:
        best = -math.inf
        for i in range(len(structure[depth][node_index])):
            val = minimax(depth + 1, structure[depth][node_index][i], False, values, alpha, beta, structure)
            best = max(best, val)
            alpha = max(alpha, best)
            if beta <= alpha:
                break
        return best
    
    else:
        best = math.inf
        for i in range(len(structure[depth][node_index])):
            val = minimax(depth + 1, structure[depth][node_index][i], True, values, alpha, beta, structure)
            best = min(best, val)
            beta = min(beta, best)
            if beta <= alpha:
                break
        return best

values = [5, 7, 3, 9, 2, 6]
structure = {
    0: {0: [0, 1, 2]}, 
    1: {0: [0, 1], 1: [2, 3, 4], 2: [5]} 
}

optimal_value = minimax(0, 0, True, values, -math.inf, math.inf, structure)

print(f"Optimal Value for MAX: {optimal_value}")
```
