```python
import math

def minimax(scores, depth, node_index, is_max, alpha, beta):
    if depth == 3:
        return scores[node_index]
    if is_max:
        best = -math.inf
        for i in range(2):
            val = minimax(scores, depth + 1, node_index * 2 + i, False, alpha, beta)
            best = max(best, val)
            alpha = max(alpha, best)
            if beta <= alpha:
                break
        return best
    else:
        best = math.inf
        for i in range(2):
            val = minimax(scores, depth + 1, node_index * 2 + i, True, alpha, beta)
            best = min(best, val)
            beta = min(beta, best)
            if beta <= alpha:
                break
        return best

left_scores = [10, 15, 5, 8]
right_scores = [12, 6, 9, 3]

left_value = minimax(left_scores, 1, 0, False, -math.inf, math.inf)
right_value = minimax(right_scores, 1, 0, False, -math.inf, math.inf)

optimal_value = max(left_value, right_value)
print("Optimal value:", optimal_value)
```
